```python
coin_id = "bitcoin"
url = f"https://api.coingecko.com/api/v3/coins/{coin_id}/market_chart"
params = {
    "vs_currency": "usd",
    "days": 90,
    "interval": "daily"
}

response = requests.get(url, params=params, timeout=30)
data = response.json()

print("Keys:", data.keys())
print("Price sample:", data["prices"][:3])
print("Volume sample:", data["total_volumes"][:3])
print("Market cap sample:", data["market_caps"][:3])
print("Total days:", len(data["prices"]))
```
